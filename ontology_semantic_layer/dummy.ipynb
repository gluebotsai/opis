```python
import os
import re
import requests
from bs4 import BeautifulSoup


# Your Google Sheet URLs



# Folder to save files
download_folder = "google_sheets"
os.makedirs(download_folder, exist_ok=True)


def get_sheet_id(value):
    """Extract Google Sheet ID from URL or return ID directly."""

    value = value.strip()

    # Full Google Sheets URL
    match = re.search(r"/spreadsheets/d/([a-zA-Z0-9_-]+)", value)

    if match:
        return match.group(1)

    # If the value itself is a Sheet ID
    if re.fullmatch(r"[a-zA-Z0-9_-]+", value):
        return value

    return None


def get_sheet_title(sheet_id):
    """Get Google Sheet title."""

    url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/edit"

    response = requests.get(url, timeout=30)

    if response.status_code != 200:
        return None

    # Google Sheet title is usually inside <title>
    soup = BeautifulSoup(response.text, "html.parser")

    title = soup.title.string if soup.title else None

    if title:
        # Google usually returns:
        # "My Sheet - Google Drive"
        # or "My Sheet - Google Sheets"
        title = re.sub(r"\s*-\s*Google (Drive|Sheets).*$", "", title)

    return title.strip() if title else None


def clean_filename(filename):
    """Remove characters that are invalid in filenames."""

    return re.sub(r'[<>:"/\\|?*]', "_", filename)


def download_google_sheet(value, index):

    try:
        # -----------------------------
        # Get Sheet ID
        # -----------------------------
        sheet_id = get_sheet_id(value)

        if not sheet_id:
            print(f"❌ Invalid Google Sheet URL/ID: {value}")
            return

        # -----------------------------
        # Get Sheet Title
        # -----------------------------
        title = get_sheet_title(sheet_id)

        if not title:
            print(f"❌ Could not get title for: {value}")
            return

        title = clean_filename(title)

        # -----------------------------
        # Download as Excel
        # -----------------------------
        download_url = (
            f"https://docs.google.com/spreadsheets/d/"
            f"{sheet_id}/export?format=xlsx"
        )

        response = requests.get(download_url, timeout=60)

        if response.status_code != 200:
            print(
                f"❌ Failed to download '{title}' "
                f"(HTTP {response.status_code})"
            )
            return

        # -----------------------------
        # Save using Google Sheet title
        # -----------------------------
        file_path = os.path.join(
            download_folder,
            f"{title}.xlsx"
        )

        with open(file_path, "wb") as f:
            f.write(response.content)

        print(f"✅ Downloaded: {title}.xlsx")

    except Exception as e:
        print(f"❌ Error downloading '{value}': {e}")


# Download all sheets
for i, sheet in enumerate(google_sheets, start=1):
    download_google_sheet(sheet, i)
```
